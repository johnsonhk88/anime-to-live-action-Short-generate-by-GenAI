# 03 — Test Gemini / Vertex AI Connectivity

Goal: Verify Google Gemini API is accessible and can generate text/images.

Two paths:
1. **Gemini API** (via `google-genai` or `google-generativeai`) — requires `GOOGLE_API_KEY`
2. **Vertex AI** (via GCP) — requires service account credentials + `gcloud` auth

Fill in your key or auth method below.

```python
import os
from dotenv import load_dotenv

load_dotenv("../.env")

GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY", "")
print(f"GOOGLE_API_KEY set: {'Yes' if GOOGLE_API_KEY else 'No'}")
```

## 3.1 Test Gemini via google-genai (new SDK)

```python
if GOOGLE_API_KEY:
    try:
        from google import genai
        client = genai.Client(api_key=GOOGLE_API_KEY)
        
        # List available models
        print("Available models:")
        for m in client.models.list():
            if "gemini" in m.name:
                print(f"  - {m.name}")
    except Exception as e:
        print(f"ERROR listing models: {e}")
else:
    print("Skipping: GOOGLE_API_KEY not set.")
    print("Set it in backend/.env or export GOOGLE_API_KEY=...")
```

## 3.2 Test text generation

```python
if GOOGLE_API_KEY:
    try:
        from google import genai
        client = genai.Client(api_key=GOOGLE_API_KEY)
        
        resp = client.models.generate_content(
            model="gemini-2.5-flash",
            contents="Describe a Japanese anime hero in 2 sentences."
        )
        print("Response:")
        print(resp.text)
    except Exception as e:
        print(f"ERROR: {e}")
        print("\nTroubleshooting:")
        print("  1. Is the API key valid?")
        print("  2. Is the model name correct? Try 'gemini-1.5-flash'")
        print("  3. Does the API key have billing enabled?")
```

## 3.3 Test Gemini via google-generativeai (langchain-compatible)

```python
if GOOGLE_API_KEY:
    try:
        import google.generativeai as genai
        genai.configure(api_key=GOOGLE_API_KEY)
        
        model = genai.GenerativeModel("gemini-1.5-flash")
        resp = model.generate_content("Describe a Japanese anime hero in 2 sentences.")
        print("Response:")
        print(resp.text)
        print(f"\nUsage: {resp.usage_metadata}")
    except Exception as e:
        print(f"ERROR: {e}")
```

## 3.4 Test Gemini via LangChain (how the pipeline will call it)

```python
if GOOGLE_API_KEY:
    try:
        from langchain_google_genai import ChatGoogleGenerativeAI
        
        model = ChatGoogleGenerativeAI(
            model="gemini-1.5-flash",
            google_api_key=GOOGLE_API_KEY,
            temperature=0.7,
        )
        resp = model.invoke([
            ("system", "You are a helpful assistant."),
            ("human", "Describe a Japanese anime hero in 2 sentences.")
        ])
        print("Response:")
        print(resp.content)
        print(f"\nToken usage: {resp.usage_metadata}")
    except Exception as e:
        print(f"ERROR: {e}")
```

## 3.6 Test Vertex AI (GCP)

If you have a GCP project with Vertex AI enabled, test here.

```python
# Check if gcloud is authenticated
import subprocess
try:
    result = subprocess.run(
        ["gcloud", "auth", "print-access-token"],
        capture_output=True, text=True, timeout=5
    )
    if result.returncode == 0:
        print("gcloud authenticated: Yes")
        print(f"Project: ", end="")
        subprocess.run(["gcloud", "config", "get-value", "project"])
    else:
        print(f"gcloud not authenticated. Run: gcloud auth login")
        print(f"Error: {result.stderr[:200]}")
except FileNotFoundError:
    print("gcloud CLI not found. Install Google Cloud SDK.")
except Exception as e:
    print(f"Error: {e}")
```
